## Menentukan Pertanyaan Bisnis

**SMART Question** adalah sebuah framework untuk merumuskan pertanyaan secara terstruktur agar memeperoleh informasi yang mendalam.

* **Spesific (Spesifik)**
  * Pertanyaan harus jelas, fokus pada sebuah topik tertentu, dan tidak bermakna ganda. Hindari pertanyaan yang terlalu luas.
    * Salah: Bagaimana cara meningkatkan penjualan?
    * Benar: Faktor apa saja yang memengaruhi penurunan penjualan produk kategori elektronik di wilayah Jakarta selama kuartal terakhir?
* **Measurable (Terukur)**
  * Pertanyaan harus bisa dijawab dengan angka atau matrix yang konkret, harus tahu apa yang akan dihitung.
    * Salah: Apakah pelanggan senang dengan layanan kita?
    * Benar: Berapa skor rata-rata Customer Satisfaction untuk layanan purna jual bulan ini dibandingkan bulan lalu?
* **Action-Oriented (Berorientasi Aksi)**
  * Hasil dari pertanyaan harus bisa memberikan arahan untuk melakukan tindakan nyata. Jika pertanyaan terjawab, stakeholder harus tahu apa langkah selanjutnya.
    * Salah: Mengapa orang suka berbelanja?
    * Benar: Fitur apa pada aplikasi yang paling sering digunakan sebelum pengguna memutuskan untuk melakukan checkout?
* **Relevant (Relevan)**
  * Hasil dari pertanyaan harus sejalan dengan tujuan utama bisnis atau masalah yang sedang dihadapi.
    * Salah: Menanyakan tentang stok gudang saat masalah utamanya adalah efektivitas kampanye media sosial.
    * Benar: Apakah kampanye iklan di Instagram memberikan Return on Ad Spend (ROAS) yang lebih tinggi dibandingkan iklan di TikTok?
* **Time-bound (Terikat Waktu)**
  * Pertanyaan harus ada batasan waktu yang jelas agar analisis memiliki konteks yang tepat.
    * Salah: Berapa banyak pengguna baru kita?
    * Benar: Berapa tingkat pertumbuhan pengguna baru secara bulanan (Month-over-Month) sepanjang tahun 2025?

**Contoh pertanyaan bisnis yang memenuhi seluruh elemen SMART**

***"Faktor apa saja yang memengaruhi penurunan conversion rate pada pengguna aplikasi Android di wilayah Jabodetabek sebesar 5% selama periode Flash Sale Maret 2026?"***

Keterangan:

- **Specific**: Fokus pada "penurunan conversion rate" untuk "aplikasi Android" di "Jabodetabek". Bukan sekadar penjualan turun.
- **Measurable**: Ada angka konkret yang ingin dianalisis, yaitu penurunan sebesar "5%".
- **Action-Oriented**: Dengan mengetahui faktor penyebabnya misalnya bug pada versi Android tertentu atau kendala logistik di Jabodetabek, tim bisa langsung melakukan perbaikan teknis atau operasional.
- **Relevant**: Penurunan konversi saat Flash Sale adalah masalah kritis bagi bisnis retail/e-commerce.
- **Time-bound**: Dibatasi pada periode spesifik "Flash Sale Maret 2026".

- **Pertanyaan 1:** Kategori produk apa saja yang menghasilkan total pendapatan (revenue) tertinggi dan terendah sepanjang periode transaksi tahun 2017 hingga 2018, serta bagaimana tren pertumbuhannya per bulan?
Keterangan :
- Specific: Berfokus pada identifikasi kategori produk dengan pendapatan tertinggi dan terendah beserta tren pertumbuhan bulanannya.
- Measurable: Diukur menggunakan metrik konkret berupa total pendapatan (revenue dalam satuan mata uang BRL) dan persentase/selisih pertumbuhan bulanan.
- Action-Oriented: Hasil analisis menjadi dasar keputusan tim bisnis untuk memprioritaskan alokasi stok, kampanye promosi pada kategori unggulan, serta evaluasi kategori berkinerja rendah.
- Relevant: Pendapatan per kategori produk adalah indikator utama performa penjualan pada bisnis e-commerce.
- Time-bound: Dibatasi pada rentang waktu transaksi sepanjang tahun 2017 hingga 2018.

- **Pertanyaan 2:** Bagaimana distribusi segmen pelanggan berdasarkan nilai Recency, Frequency, dan Monetary (RFM) selama tahun 2018, serta segmen mana yang memberikan kontribusi pendapatan terbesar bagi perusahaan?
Keterangan:
- Specific: Menganalisis segmentasi pelanggan berdasarkan tiga parameter perilaku transaksi (Recency, Frequency, dan Monetary) serta kontribusi pendapatannya.
- Measurable: Dihitung menggunakan jumlah hari sejak transaksi terakhir (Recency), jumlah total pesanan (Frequency), dan total nilai belanja (Monetary).
- Action-Oriented: Memungkinkan tim pemasaran merancang strategi retensi yang tepat sasaran, seperti pemberian program loyalitas bagi Best Customers atau promo reaktivasi bagi pelanggan yang berisiko hilang (At Risk).
- Relevant: Memahami segmen pelanggan bernilai tinggi sangat krusial untuk meningkatkan retensi dan efisiensi anggaran pemasaran.
- Time-bound: Dibatasi pada aktivitas transaksi pelanggan sepanjang tahun 2018.

## Data Wrangling

### Gathering Data

### Cleaning Data

#### Fixing Data Type & Missing Values

```python
# 1. Mengubah tipe data kolom tanggal menjadi datetime pada orders_df
datetime_columns = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

for column in datetime_columns:
    orders_df[column] = pd.to_datetime(orders_df[column])

# 2. Menangani missing values pada products_df (mengisi kategori kosong dengan 'unknown')
products_df['product_category_name'] = products_df['product_category_name'].fillna('unknown')

# 3. Menyaring pesanan yang valid (status 'delivered') pada tahun 2017 - 2018 sesuai pertanyaan bisnis
orders_clean_df = orders_df[
    (orders_df['order_status'] == 'delivered') &
    (orders_df['order_purchase_timestamp'].dt.year.isin([2017, 2018]))
].copy()

# 4. Menggabungkan seluruh tabel yang relevan menjadi satu DataFrame utama (all_df)
all_df = orders_clean_df.merge(order_items_df, on='order_id', how='inner')
all_df = all_df.merge(order_payments_df, on='order_id', how='inner')
all_df = all_df.merge(products_df[['product_id', 'product_category_name']], on='product_id', how='left')
all_df = all_df.merge(translation_df, on='product_category_name', how='left')
all_df = all_df.merge(customers_df[['customer_id', 'customer_unique_id', 'customer_city', 'customer_state']], on='customer_id', how='left')

# Isi nama kategori bahasa Inggris yang kosong dengan nama aslinya (termasuk 'unknown')
all_df['product_category_name_english'] = all_df['product_category_name_english'].fillna(all_df['product_category_name'])

# 5. Verifikasi hasil akhir cleaning & simpan sebagai main_data.csv untuk Streamlit
print("Tipe data tanggal:", all_df['order_purchase_timestamp'].dtype)
print("Total baris dan kolom data bersih (all_df):", all_df.shape)
all_df.to_csv("main_data.csv", index=False)
all_df.head()
```

**Insight:** (Opsional)
- Seluruh kolom tanggal pada `orders_df` telah berhasil diubah menjadi tipe `datetime`.
- Nilai kosong pada kolom `product_category_name` di `products_df` telah diisi dengan label `'unknown'`.
- Data telah disaring khusus untuk transaksi valid berstatus `'delivered'` pada rentang tahun 2017–2018 dan digabungkan menjadi satu tabel utama (`all_df`), serta diekspor ke `main_data.csv` untuk kebutuhan dashboard.

## Exploratory Data Analysis (EDA)

### Explore Revenue per Kategori Produk & Metrik RFM Pelanggan

```python
# 1. Eksplorasi Pendapatan & Jumlah Pesanan per Kategori Produk (2017-2018)
category_revenue_df = all_df.groupby("product_category_name_english").agg(
    total_orders=("order_id", "nunique"),
    total_revenue=("price", "sum")
).sort_values(by="total_revenue", ascending=False).reset_index()

print("=== TOP 10 KATEGORI PRODUK BERDASARKAN REVENUE (2017-2018) ===")
print(category_revenue_df.head(10))

print("\n=== BOTTOM 5 KATEGORI PRODUK BERDASARKAN REVENUE (2017-2018) ===")
print(category_revenue_df.tail(5))

# 2. Eksplorasi Data Pelanggan Tahun 2018 untuk Analisis RFM
df_2018 = all_df[all_df['order_purchase_timestamp'].dt.year == 2018].copy()
recent_date = df_2018['order_purchase_timestamp'].max()

rfm_df = df_2018.groupby(by="customer_unique_id", as_index=False).agg({
    "order_purchase_timestamp": lambda x: (recent_date - x.max()).days, # Recency
    "order_id": "nunique",                                             # Frequency
    "price": "sum"                                                     # Monetary
})
rfm_df.columns = ["customer_id", "recency", "frequency", "monetary"]

print("\n=== RANGKUMAN STATISTIK RFM PELANGGAN (2018) ===")
print(rfm_df.describe())
```

**Insight:**
- **Performa Kategori Produk (2017–2018):** Kategori `health_beauty`, `watches_gifts`, dan `bed_bath_table` mendominasi total pendapatan tertinggi, sementara kategori seperti `security_and_services` mencatatkan pendapatan terendah.
- **Perilaku Pelanggan 2018 (RFM):** Mayoritas pelanggan hanya melakukan 1 kali transaksi (`frequency` = 1), dengan rata-rata nilai belanja (`monetary`) sekitar 137–140 BRL dan rentang jarak transaksi terakhir (`recency`) bervariasi dari 0 hingga lebih dari 200 hari.

### Pertanyaan 2: Bagaimana segmentasi dan karakteristik pelanggan pada tahun 2018 berdasarkan analisis RFM (Recency, Frequency, Monetary)?

```python
fig, ax = plt.subplots(nrows=1, ncols=3, figsize=(22, 6))

# Top 5 Recency (Paling Baru Bertransaksi)
sns.barplot(
    y="recency",
    x="customer_id",
    data=rfm_df.sort_values(by="recency", ascending=True).head(5),
    color="#2b5c8f",
    ax=ax[0]
)
ax[0].set_ylabel("Hari sejak transaksi terakhir", fontsize=11)
ax[0].set_xlabel("Customer ID (5 Karakter Awal)", fontsize=11)
ax[0].set_title("By Recency (days)", loc="center", fontsize=14)
ax[0].set_xticklabels([x[:5] for x in rfm_df.sort_values(by="recency", ascending=True).head(5)["customer_id"]])

# Top 5 Frequency (Paling Sering Bertransaksi)
sns.barplot(
    y="frequency",
    x="customer_id",
    data=rfm_df.sort_values(by="frequency", ascending=False).head(5),
    color="#2b5c8f",
    ax=ax[1]
)
ax[1].set_ylabel("Jumlah Transaksi", fontsize=11)
ax[1].set_xlabel("Customer ID (5 Karakter Awal)", fontsize=11)
ax[1].set_title("By Frequency", loc="center", fontsize=14)
ax[1].set_xticklabels([x[:5] for x in rfm_df.sort_values(by="frequency", ascending=False).head(5)["customer_id"]])

# Top 5 Monetary (Total Belanja Terbesar)
sns.barplot(
    y="monetary",
    x="customer_id",
    data=rfm_df.sort_values(by="monetary", ascending=False).head(5),
    color="#2b5c8f",
    ax=ax[2]
)
ax[2].set_ylabel("Total Belanja (BRL)", fontsize=11)
ax[2].set_xlabel("Customer ID (5 Karakter Awal)", fontsize=11)
ax[2].set_title("By Monetary (BRL)", loc="center", fontsize=14)
ax[2].set_xticklabels([x[:5] for x in rfm_df.sort_values(by="monetary", ascending=False).head(5)["customer_id"]])

plt.suptitle("Best Customers Based on RFM Parameters (2018)", fontsize=18)
plt.tight_layout()
plt.show()
```

- **Visualisasi Pertanyaan 2 (RFM):** Pelanggan terbaik di tahun 2018 memiliki nilai *recency* 0 hari (bertransaksi tepat di hari terakhir pencatatan data), frekuensi pembelian maksimal mencapai 9–15 kali transaksi, serta nilai belanja individu tertinggi menembus lebih dari 13.000 BRL.

## Analisis Lanjutan (Opsional)

```python
# Segmentasi Pelanggan berdasarkan Skor RFM (2018)
rfm_df['r_rank'] = rfm_df['recency'].rank(ascending=False)
rfm_df['f_rank'] = rfm_df['frequency'].rank(ascending=True)
rfm_df['m_rank'] = rfm_df['monetary'].rank(ascending=True)

# Normalisasi skor ke skala 0 - 100
rfm_df['r_rank_norm'] = (rfm_df['r_rank'] / rfm_df['r_rank'].max()) * 100
rfm_df['f_rank_norm'] = (rfm_df['f_rank'] / rfm_df['f_rank'].max()) * 100
rfm_df['m_rank_norm'] = (rfm_df['m_rank'] / rfm_df['m_rank'].max()) * 100

# Menghitung skor RFM tertimbang (Skala 0 - 5)
rfm_df['RFM_score'] = (
    0.15 * rfm_df['r_rank_norm'] +
    0.28 * rfm_df['f_rank_norm'] +
    0.57 * rfm_df['m_rank_norm']
) * 0.05
rfm_df['RFM_score'] = rfm_df['RFM_score'].round(2)

# Mengelompokkan pelanggan ke dalam segmen bisnis
def segment_customer(score):
    if score > 4.5:
        return "Top Customers"
    elif score > 4.0:
        return "High Value Customers"
    elif score > 3.0:
        return "Medium Value Customers"
    elif score > 1.6:
        return "Low Value Customers"
    else:
        return "Lost / Dormant Customers"

rfm_df["customer_segment"] = rfm_df["RFM_score"].apply(segment_customer)

# Visualisasi distribusi jumlah pelanggan per segmen
segment_counts = rfm_df["customer_segment"].value_counts().reset_index()
segment_counts.columns = ["customer_segment", "customer_count"]

plt.figure(figsize=(10, 5))
sns.barplot(
    x="customer_count",
    y="customer_segment",
    data=segment_counts,
    palette="Blues_r"
)
plt.title("Distribusi Segmen Pelanggan E-Commerce (2018)", fontsize=14)
plt.xlabel("Jumlah Pelanggan", fontsize=11)
plt.ylabel("Segmen Pelanggan", fontsize=11)
plt.tight_layout()
plt.show()

print(segment_counts)
```

**Insight:**
- Melalui pemeringkatan RFM tertimbang (*weighted RFM score*), basis pelanggan tahun 2018 berhasil diklasifikasikan ke dalam lima segmen perilaku: *Top Customers*, *High Value Customers*, *Medium Value Customers*, *Low Value Customers*, dan *Lost / Dormant Customers*.
- Mayoritas pelanggan berada pada segmen *Low Value* dan *Medium Value Customers* karena karakteristik pembelian satu kali (*one-time buyer*), sedangkan porsi *Top Customers* dan *High Value Customers* merupakan kelompok kecil namun memberikan kontribusi pendapatan terbesar.
